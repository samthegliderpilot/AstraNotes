```python
#%%
import sys
from pathlib import Path

# Assume the notebook is in project_root/notebooks/
# This adds project_root/ to the path
sys.path.append(str(Path.cwd().parent))

from astrocalc.keplerian_ui import OrbitalMechanicsWidget
from astrocalc.equations.keplerian_equations import KeplerianEquations
from astrocalc.util.equation_helpers import create_equation_renderers, EquationGroup, render_equation_groups
import ipywidgets as widgets
from IPython.display import display, Math, Markdown

# this might be needed for better page formatting (especally as things grow)
from IPython.display import display, HTML
display(HTML("<style>.container { width:95% !important; }</style>"))
orbital = KeplerianEquations()
# Equations
equation_groups = [
    EquationGroup("General Orbital Equations", [
        orbital.vis_viva(),
        orbital.mean_motion(),
        orbital.orbital_period(),
    ]),
    EquationGroup("Circular Orbit Equations", [
        orbital.circular_velocity(),
    ]),
    EquationGroup("Escape Conditions", [
        orbital.escape_velocity(),
    ]),
]

# Create renderers once
equation_renderers = create_equation_renderers(equation_groups)

# Render UI
eq_grid = render_equation_groups(equation_groups, equation_renderers)

# Create the input widget system
elements_widget = OrbitalMechanicsWidget(orbital, equation_renderers)

# Define how to evaluate and update all equation displays
def evaluate_all_equations(b):
    elements_widget.evaluate_and_display(equation_renderers)

# Connect the evaluate button in the widget system
elements_widget.evaluate_button.on_click(evaluate_all_equations)

# Build the equation display grid


# Display everything
display(
    widgets.HTML("<h3>Orbital Elements</h3>"),
    elements_widget.unit_selectors,
    elements_widget.a_widget,
    elements_widget.e_float_widget,
    elements_widget.i_widget,
    elements_widget.raan_widget,
    elements_widget.arg_pe_widget,
    elements_widget.nu_widget,
    elements_widget.mu_widget,
    elements_widget.evaluate_button,
    widgets.HTML("<h3>Equations</h3>"),
    render_equation_groups(equation_groups, equation_renderers)
)
```
