```python
#%%
from astrocalc.keplerian_ui import OrbitalMechanicsWidget
from astrocalc.cheatsheet.keplerian_equations import KeplerianEquations
from astrocalc.util.equation_helpers import create_equation_renderers, EquationGroup, render_equation_groups

import ipywidgets as widgets
from IPython.display import display, HTML

display(HTML("<style>.container { width:95% !important; }</style>"))
from IPython.display import HTML, display

display(HTML("""
<style>
/* Highlight invalid ipywidgets inputs */
.astro-invalid input {
  background: #ffe6ea !important;   /* light pink */
  border: 2px solid #d33 !important;
  border-radius: 4px;
}

/* Optional: slightly emphasize the label text too (FloatText description) */
.astro-invalid .widget-label {
  color: #b00 !important;
  font-weight: 600;
}
</style>
"""))

orbital = KeplerianEquations()

equation_groups = [
    EquationGroup("General Orbital Equations", [
        orbital.vis_viva(),
        orbital.mean_motion(),
        orbital.orbital_period(),
        orbital.semi_latus_rectum(),
        orbital.velocity_elliptical(),
        orbital.velocity_elliptical(),
    ]),
    EquationGroup("Elliptical Orbtail Equations",
                  [
                      orbital.sin_eccentric_anomaly_wrt_true_anomaly(),
                      orbital.cos_eccentric_anomaly_wrt_true_anomaly(),
                      orbital.eccentric_anomaly_wrt_true_anomaly()
                  ]
    ),
    EquationGroup("Circular Orbit Equations", [
        orbital.circular_velocity(),
    ]),
    EquationGroup("Parabolic Equations", [
        orbital.escape_velocity(),
    ]),
]

equation_renderers = create_equation_renderers(equation_groups)
elements_widget = OrbitalMechanicsWidget(orbital, equation_renderers)

def evaluate_all_equations(_b=None):
    elements_widget.evaluate_and_display()

elements_widget.evaluate_button.on_click(evaluate_all_equations)

# Initial evaluation
elements_widget.evaluate_and_display()

display(
    widgets.HTML("<h3>Orbital Elements</h3>"),
    elements_widget.unit_selectors,
    *elements_widget.get_field_container_list(),
    elements_widget.evaluate_button,
    widgets.HTML("<h3>Orbit Diagram</h3>"),
    elements_widget.orbit_diagram_out,    
    widgets.HTML("<h3>Equations</h3>"),
    render_equation_groups(equation_groups, equation_renderers),
)
```

```python
import ipywidgets as widgets
from astrocalc.cheatsheet.jupyter_sources_render import render_sources_jupyter

sources_widget = render_sources_jupyter(equation_groups, include_group_name=True)

acc = widgets.Accordion(children=[sources_widget])
acc.set_title(0, "Sources")
acc.selected_index = None  # collapsed by default

display(acc)
```
