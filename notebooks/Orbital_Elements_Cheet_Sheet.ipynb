```python
#%%
from astranotes.keplerian_ui import OrbitalMechanicsWidget
from astranotes.cheatsheet.keplerian_equations import KeplerianEquations
from astranotes.util.equation_helpers import create_equation_renderers, EquationGroup, render_equation_groups

import ipywidgets as widgets
from IPython.display import display, HTML

display(HTML("<style>.container { width:95% !important; }</style>"))
from IPython.display import HTML, display

display(HTML("""
<style>
/* Highlight invalid ipywidgets inputs */
.astro-invalid input {
  background: #ffe6ea !important;   /* light pink */
  border: 2px solid #d33 !important;
  border-radius: 4px;
}

/* Optional: slightly emphasize the label text too (FloatText description) */
.astro-invalid .widget-label {
  color: #b00 !important;
  font-weight: 600;
}
</style>
"""))

orbital = KeplerianEquations()

equation_groups = [
    EquationGroup("General Orbital Equations", [
        orbital.vis_viva,
        orbital.mean_motion,
        orbital.orbital_period,
        orbital.semi_latus_rectum,
        orbital.velocity_magnitude,
        orbital.radius_of_periapsis,
        orbital.radius_of_apoapsis,
        orbital.flight_path_angle_wrt_eccentric_anomaly,
        orbital.angular_momentum,
        orbital.circular_velocity,
    ]),
    EquationGroup("Elliptical Orbtail Equations",
                  [
                      orbital.eccentric_anomaly_wrt_true_anomaly,
                      orbital.mean_anomaly_ellitpical
                  ]
    ),
    EquationGroup("Parabolic Orbit Equations", [
        
        orbital.parabolic_anomaly_wrt_true_anomaly,
        orbital.flight_path_angle_parabolic,
        orbital.semi_latus_rectum_parabolic
    ]),
    EquationGroup("Hyperbolic Equations", [
        orbital.escape_velocity,
        orbital.hyperbolic_anomaly_wrt_true_anomaly,
        orbital.mean_anomaly_hyperbolic
    ]),
]

equation_renderers = create_equation_renderers(equation_groups)
elements_widget = OrbitalMechanicsWidget(orbital, equation_renderers)

def evaluate_all_equations(_b=None):
    elements_widget.evaluate_and_display()

elements_widget.evaluate_button.on_click(evaluate_all_equations)

# Initial evaluation
elements_widget.evaluate_and_display()

display(
    widgets.HTML("<h3>Orbital Elements</h3>"),
    elements_widget.unit_selectors,
    *elements_widget.get_field_container_list(),
    elements_widget.evaluate_button,
    widgets.HTML("<h3>Orbit Diagram</h3>"),
    elements_widget.orbit_diagram_out,    
    widgets.HTML("<h3>Equations</h3>"),
    render_equation_groups(equation_groups, equation_renderers),
)
```

```python
import ipywidgets as widgets
from astranotes.cheatsheet.jupyter_sources_render import render_sources_jupyter
from datetime import datetime, timezone
import importlib.metadata as md


version = md.version("astranotes")
stamp = f"Version {version} — Generated {datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M UTC')}"

atan2_note = (
    "<b>Note on atan2:</b> This sheet uses <code>atan2(y, x)</code> "
    "(sine term / <i>y</i> first, cosine term / <i>x</i> second) to preserve quadrant."
)

sources_widget = render_sources_jupyter(
    equation_groups,
    include_group_name=False,
    stamp_html=stamp,
    footer_html=atan2_note,
)

acc = widgets.Accordion(children=[sources_widget])
acc.set_title(0, "Sources and Notes")
acc.selected_index = None  # collapsed by default

display(acc)

```
